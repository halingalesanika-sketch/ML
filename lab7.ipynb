```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
data = pd.read_csv("/content/Pass-Fail Data.csv")
```

```python
print("First 5 records:")
print(data.head())
```

```python
print("\nDataset Information:")
print(data.info())
```

```python
print("\nColumn Names:")
print(data.columns)
```

```python
print("\nMissing Values:")
print(data.isnull().sum())
```

```python
print("\nDuplicate Values:")
print(data.duplicated().sum())
```

```python
X = data[
    [ "attendance_pct",
     "homework_pct",
      "midterm_score",
      "study_hours_per_week"
      ]
    ]
y = data["pass"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42 )
print("\nTraining data size:", X_train.shape)
print("Testing data size:", X_test.shape)
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print("\nActual Values:")
print(y_test.values)
print("\nPredicted Values:")
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy)
print("Accuracy Percentage:", accuracy * 100, "%")
```

```python
cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)
```

```python
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
new_student = [[85, 80, 75, 10]]

prediction = model.predict(new_student)
print("\nNew Student Prediction:")
if prediction[0] == 1:
  print("Student will PASS")
else:
  print("Student will FAIL")
```

```python
probability = model.predict_proba(new_student)

print("\nPrediction Probability:")
print("Fail Probability:", probability[0][0])
print("Pass Probability:", probability[0][1])
```

```python
print("\nModel Coefficients:")

for column, coefficient in zip(X.columns, model.coef_[0]):
  print(column, ":", coefficient)

print("\nIntercept:", model.intercept_[0])
```
